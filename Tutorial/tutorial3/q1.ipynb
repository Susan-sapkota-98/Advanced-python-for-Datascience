#3. Create a Rectangle class with attributes length and width . Add methods to calculate area and perimeter.

```python
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width

    def area(self):
        return self.length*self.width
    
    def perimeter(self):
        return 2*(self.length+self.width)



res=Rectangle(10,20)
print("Length: ",res.length)
print("width: ",res.width)
print("The area of rectangle is :  ",res.area())
print("The perimeter of rectangle is : ",res.perimeter())
```

#Create a Circle class with attribute radius . Use encapsulation to protect radius and add getter and setter methods. Add a method to calculate the area.

```python
class Circle:
    def __init__(self,radius):
        self.radius=radius

    def getter(self):
        return self.radius
    
    def setter(self,radius):
        self.radius=radius

    def area(self):
        return 3.14 * self.radius *self.radius
    
c=Circle(2)

print("The radius is: ",c.radius)
print('The area of circle is: ',c.area())

c.setter(5)

print("The radius is: ",c.radius)
print('The area of circle is: ',c.area())
```

#5. Create a BankAccount class. Add methods for deposit, withdraw, and checking balance. Ensure the balance cannot be accessed directly.

```python
class BankAccount:
    def __init__(self,balance):
        self.balance=balance

    def deposit(self,amount):
        self.balance +=amount
        print("The deposited amount is ",amount)
        
    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance -=amount
            print("The Withdraw amount is : ",amount)
        else:
            print("Insufficient balance")

    def checkBalance(self):
        return self.balance
acc=BankAccount(1000)
print("The actual balance is : ",acc.balance)
acc.deposit(1000)
acc.withdraw(500)
print("The current balance is ", acc.checkBalance())
```

#6. Create a Student class with attributes name , roll_number , and marks . Add methods to display student info and calculate average marks.

```python
class Student:
    def __init__(self,name,roll_number,marks):
        self.name=name
        self.roll_number=roll_number
        self.marks=marks

    def display_inf(self):
        print('Name:',self.name)
        print('roll_number:',self.roll_number)
        print('marks:',self.marks)
    

    def avg(self):
        return sum(self.marks)/len(self.marks)
    
s1=Student("Ram",1,[10,20,30,40,50])
s1.display_inf()
print("The average marks of the subject: ", s1.avg())
```

#7. Implement a Teacher class that inherits from Person . Override the greet method to include subject taught.

```python
class Person():
    def __init__(self,name):
        self.name=name

    def greet(self):
        print("Hello my name is ",self.name)

class Teacher(Person):
    def __init__(self, name,subject):
        super().__init__(name)
        self.subject=subject

    def greet(self):
        print("Hello, my name is ",self.name, "and I teach ",self.subject)

t1=Teacher("Ram","Math")
t1.greet()
```

#8. Create a LibraryBook class with attributes title , author , available . Add methods to borrow and return the book. Use encapsulation for available .

```python
class LibraryBook:
    def __init__(self,title,author,available=True):
        self.title=title
        self.author=author
        self.available=available

    def borrow(self):
        if self.available:
            self.available=False
            print("You borrowed the book:",self.title)

        else:
            print("Sorry, the book is not available")

    def retur_book(self):
        self.available=True
        print("You returned the book:",self.title)

    def is_available(self):
        return self.available
    
book1=LibraryBook("python","John")

book1.borrow()
print("Available: ",book1.is_available())

book1.retur_book()
print("Available: ",book1.is_available())
```

#9. Create a Shape class and two subclasses Triangle and Square . Implement area calculation for both using inheritance.

```python
class Shape:
    def area(self):
        pass


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side


t = Triangle(10, 5)
s = Square(4)

print("Area of Triangle:", t.area())
print("Area of Square:", s.area())
```

#10. Implement a CompanyEmployee class. Include name , salary , and position . Add methods to display info and give a raise. Demonstrate encapsulation for salary.

```python
class CompanyEmployee:
    def __init__(self, name, salary, position):
        self.name = name
        self.__salary = salary   # private attribute
        self.position = position

    
    def display_info(self):
        print("Name:", self.name)
        print("Position:", self.position)
        print("Salary:", self.__salary)

    # Method to give raise
    def give_raise(self, amount):
        self.__salary += amount
        print("Salary increased by:", amount)

    # Getter for salary
    def get_salary(self):
        return self.__salary


# Create object
emp = CompanyEmployee("Ram", 50000, "Manager")

emp.display_info()

emp.give_raise(5000)

print("Updated Salary:", emp.get_salary())
```
