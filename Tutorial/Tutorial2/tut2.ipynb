```python
while True:
    try:
        num=next(iter_numbers)
        print(num)
    except StopIteration:
        break
```

```python
class CountUp:
    def __init__(self,limit):

        self.limit=limit
        self.count=0

        def __iter__(self):
            return self
        
        def __next__(self):
            if self.current < self.limit:
                self.current +=1
                return self.current
            else:
                raise StopIteration
            
counter=CountUp(5)
for num in counter:
    print(num)
```

#Generator
It is a special function that returns an iterator using yield

```python
def squares(n):
    for i in range(n):
        yield i*i

for val in squares(5):
    print(val)
```

#Decorator
A decorator is a function that takes another function and extends or modifies its behavior without changing the code

```python
#Logging decorator
def logger(func):
    def wrapper(*args,**kwargs):
        print(f'calling {func.__name__} with {args} {kwargs}')
        result=func(*args,**kwargs)
        print(f'Result={result}')
        return result
    return wrapper

@logger
def add(a,b):
    return a+b
print("**")
add(3,4)
```

```python
def fibo(n):
    a,b=0,1
    for _ in range(n):
        yield a
        a,b=b,a+b

for num in fibo(7):
    print(num)
```

#decorator

```python
def decorator(func):
    def wrapper():
        print('start')
        func()
        print('end')
    return wrapper

@decorator
def greet():
    print("hello")
greet()
```

```python
import time

def timedecorator(func):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=func(*args,**kwargs)
        end=time.time()
        print('time:',end-start)
        return result
    return wrapper

@timedecorator
def hello():
    for _ in range(10000):
        pass

hello()
```

#OOP

```python
class Animal:
    def __init__(self,name):
        self.name=name

    def speaks(self):
        print(f'{self.name} is speaking')

        
```

```python
class dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed
    def speaks(self):
        print(f'{self.name} is barking')
```

```python
d1=dog('Lucky','Shepherd')
d1.speaks()
```
