#6.Recreate the same schema using SQLAlchemy ORM.

```python
from sqlalchemy import create_engine
engine = create_engine("sqlite:///library.db")
```

```python
from sqlalchemy.orm import declarative_base
from sqlalchemy import Column, Integer, String,Boolean
Base = declarative_base()
class Library(Base):
    __tablename__ = "Library"
    BookID = Column(Integer, primary_key=True)
    Title=Column(String)
    Author = Column(String)
    PublishedYear = Column(Integer)
    isAvailable=Column(Boolean, default=1)
Base.metadata.create_all(engine)
```

#adding records

```python
from sqlalchemy.orm import sessionmaker

Session = sessionmaker(bind=engine)
session = Session()

# Convert tuples into Library objects
books_objects = [
    Library(BookID=101, Title="Python", Author="John", PublishedYear=2000, isAvailable=True),
    Library(BookID=102, Title="C", Author="Ram", PublishedYear=2000, isAvailable=False),
    Library(BookID=103, Title="OOP", Author="Susan", PublishedYear=2002, isAvailable=True),
    Library(BookID=104, Title="JAVA", Author="Peter", PublishedYear=2001, isAvailable=True),
    Library(BookID=105, Title="JavaScript", Author="John", PublishedYear=2005, isAvailable=True),
    Library(BookID=106, Title="Math", Author="Shyam", PublishedYear=2010, isAvailable=False)
]


session.add_all(books_objects)


session.commit()
```

#Querying Records

```python
books = session.query(Library).all()
for b in books:
    print(b.BookID, b.Title, b.Author,b.PublishedYear,b.isAvailable)
```

#Updating Records

```python
# Query the book with BookID 101
book = session.query(Library).filter_by(BookID=101).first()

# Update availability
book.isAvailable = False  

session.commit()
```

```python
book = session.query(Library).filter_by(BookID=101).first()
print(book.BookID, book.Title, book.isAvailable)
```

#Deleting Records

```python
session.delete(book)
session.commit()
```

```python
books = session.query(Library).all()
for b in books:
    print(b.BookID, b.Title, b.Author,b.PublishedYear,b.isAvailable)
```
