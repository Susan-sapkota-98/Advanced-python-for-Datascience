#1.Create a SQLite Database for a library system.


```python
import sqlite3
conn=sqlite3.connect("library_system.db")
cursor=conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE Books (
    BookID INTEGER PRIMARY KEY,
    Title TEXT,
    Author TEXT,
    PublishedYear INTEGER,
    isAvailable BOOLEAN DEFAULT 1
);
""")
conn.commit()
```

#2.Insert at least five book records.

```python
books = [
    (101, "Python", "John", 2000,True),
    (102, "C", "Ram", 2000,False),
    (103, "OOP", "Susan", 2002,True),
    (104, "JAVA", "Peter", 2001,True),
    (105, "JavaScript", "John", 2005,True),
    (106, "Math", "Shyam", 2010,False)
]
```

```python
cursor.executemany(
    "INSERT INTO Books (BookID, Title, Author, PublishedYear,isAvailable) VALUES (?, ?, ?, ?,?)",
    books
)
conn.commit()
```

#3. Write queries to retrieve books by author.

```python
cursor.execute("SELECT * FROM Books WHERE Author='John'")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

#4. Update book availability status

```python
cursor.execute("""
UPDATE Books
SET isAvailable = 0
WHERE BookID = 101
""")
conn.commit()
```

```python
cursor.execute("SELECT BookID, Title,Author,PublishedYear, isAvailable FROM Books WHERE BookID = 101")
print(cursor.fetchone())
```

```python
cursor.execute(
    "DELETE FROM Books WHERE Title=?",("JAVA",)
)
conn.commit()
```
