```python
import pandas as pd
import numpy as np

# Sample dataset with missing values
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', None, 'David', 'Emma'],
    'Age': [25, None, 30, 22, None],
    'City': ['Kathmandu', 'Pokhara', 'Lalitpur', None, 'Bhaktapur']
})

print("Original DataFrame:")
print(data)
```

#Remove rows with any missing values

```python
clean_data = data.dropna()
print("\nDataFrame after removing rows with missing values:")
print(clean_data)
```

#Exercise 2: Replace missing values using mean or median

```python
# Fill missing values in 'Age' with mean
data['Age_mean_filled'] = data['Age'].fillna(data['Age'].mean())

# Fill missing values in 'Age' with median
data['Age_median_filled'] = data['Age'].fillna(data['Age'].median())

print("\nAfter filling missing values:")
print(data)
```

#Exercise 3: Label encoding for categorical column

```python
from sklearn.preprocessing import LabelEncoder

# Sample categorical column
data_cat = pd.DataFrame({
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

le = LabelEncoder()
data_cat['Department_encoded'] = le.fit_transform(data_cat['Department'])

print("\nLabel encoded column:")
print(data_cat)
```

#Exercise 4: One-hot encoding

```python
# One-hot encoding using pandas
data_onehot = pd.get_dummies(data_cat, columns=['Department'])
print("\nOne-hot encoded DataFrame:")
print(data_onehot)
```

#Exercise 5: Min-Max normalization

```python
# Min-Max normalization for Age column
data['Age_normalized'] = (data['Age'] - data['Age'].min()) / (data['Age'].max() - data['Age'].min())
print("\nMin-Max normalized Age:")
print(data[['Age', 'Age_normalized']])
```

#Exercise 6: Standardization (z-score)

```python
# Standardize using (x - mean)/std
data['Age_standardized'] = (data['Age'] - data['Age'].mean()) / data['Age'].std()
print("\nStandardized Age:")
print(data[['Age', 'Age_standardized']])
```
