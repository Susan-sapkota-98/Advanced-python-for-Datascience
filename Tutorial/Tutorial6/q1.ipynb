#1. Install required libraries and verify that imports run without errors.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

#2. Send a GET request to a public webpage and print the response status code.

```python
url="https://www.w3schools.com/html/html_tables.asp"
response=requests.get(url)

print(response.status_code)
```

#3. Print the first portion of the HTML content returned by the server.

```python
print(response.text[:500])
```

#4. Parse the page and count how many tables exist.

```python
# soup=BeautifulSoup(response.text,"html.parser")
soup=BeautifulSoup(response.text,"lxml")
```

```python
print(len(soup.find_all("table")))
```

#5. Extract the page title and display it.

```python
title=soup.find("title")
print(title)
```

#6. Retrieve all hyperlinks from the page.

```python
for link in soup.find_all("a"):
    href = link.get("href")
    if href:
        print(href)
```

#7. Extract one table and store rows in a Python list.

```python
table=soup.find('table')
rows=table.find_all("tr")

list=[]
for row in rows:
    cols=row.find_all(["td","th"])
    cols=[col.text.strip() for col in cols]
    list.append(cols)
print(list)
```

#8. Separate table headers from data rows.

```python
headers = [th.text.strip() for th in table.find_all("th")]

data_rows = []
for row in table.find_all("tr")[1:]:
    cols = [td.text.strip() for td in row.find_all("td")]
    data_rows.append(cols)
print(data_rows)
```

#9. Convert extracted data into a pandas DataFrame.

```python
df=pd.DataFrame(data_rows[1:],columns=data_rows[0])
print(df.head())
```

#10. Save the DataFrame to CSV format.

```python
df.to_csv('table.csv',index=False)
```

#11. Clean whitespace and newline characters from all cells.

```python
df=df.applymap(lambda x:x.strip() if isinstance(x,str) else x)
print(df.head())
```
