#1. Merge two datasets:

```python
import pandas as pd

students=pd.DataFrame({
    'id':[1,2,3,4,5],
    'name':['Alice','Bob','Charlie','David','Emma']
})

grades=pd.DataFrame({
    'id':[1,2,3,5,6],
    'score':[85,90,78,88,92]
})
```

```python
merged=pd.merge(students,grades,on=['id'])
print(merged)
```

#2. Convert a wide dataset into long format using melt().

```python
melted = pd.melt(merged, id_vars=['id'], value_vars=['name','score'], 
                 var_name='attribute', value_name='value')

print(melted)
```

#4. Use groupby() to calculate total revenue per region.

```python
import pandas as pd

data = pd.DataFrame({
    'Region': ['East', 'West', 'East', 'North', 'West', 'East'],
    'Revenue': [1000, 1500, 2000, 1200, 1800, 2500]
})

print(data)
```

```python
total_revenue = data.groupby('Region')['Revenue'].sum().reset_index()
print(total_revenue)
```

#5. Normalize sales values within each product category using transform().

```python
import pandas as pd

data = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    'Product': ['P1', 'P2', 'P3', 'P4', 'P5', 'P6'],
    'Sales': [100, 200, 300, 400, 500, 600]
})

print(data)
```

```python
data['Sales_Normalized'] = data.groupby('Category')['Sales'].transform(
    lambda x: (x - x.min()) / (x.max() - x.min())
)

print(data)
```
