```python
import matplotlib.pyplot as plt
import numpy as np

X=np.arange(1,10,0.2)
y=np.square(X)
y_cube=np.power(X,3)


plt.plot(X,y,color='r',label='square')
plt.plot(X,y_cube,color='g',label='cube')

plt.xlabel("Time")
plt.ylabel("Sales")
plt.title("sales over time")
plt.legend()
plt.show()

```

```python
import matplotlib.pyplot as plt

categories=['Apple','banana','cherries','Dates']
values=[10,20,7,12]

plt.bar(x=categories,height=values,width=0.4,edgecolor='black',color=['r','b','y','g'],alpha=0.5)


plt.xlabel('Fruit')
plt.ylabel('Quantity')
plt.xticks(rotation=45)
plt.title('Fruit Inventory')

plt.show()
```

#histogram

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

data=np.random.normal(loc=50,scale=10,size=1000)

# plt.hist(data,bins=20,color='skyblue',kde='line',edgecolor='black')
sns.histplot(data,bins=20,color='red',kde='line',edgecolor='black')

plt.xlabel('value')
plt.ylabel('Frequency')
plt.title('Histogram of random data')
plt.show()
```

```python
from sklearn.datasets import make_blobs
X,y=make_blobs(
    n_features=2,n_samples=1000,random_state=42
)
print(X.shape)
```

```python
plt.scatter(X[:, 0],X[:,1],c=y,cmap='viridis',edgecolors='black')
plt.xlabel('feature 1')
plt.ylabel('feature 2')
plt.title('make blobs visualization')
plt.colorbar(label='Cluster')
plt.show()

```

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

X, y = make_blobs(
    n_features=3, 
    n_samples=1000, 
    centers=4,
    random_state=42
)

print(X.shape)  


fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=y, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
```
