#innerjoin

```python
import numpy as np
import pandas as pd
df_students=pd.DataFrame({
    'studentid':[1,2,3],
    'name':['Ram','Shyam','Hari']

})
df_marks=pd.DataFrame({
    'studentid':[1,2,4],
    'marks':[20,12,32]
})

merged_inner=pd.merge(df_students,df_marks,on='studentid',how='inner')
print(merged_inner)
```

#1. Write an iterator that returns the first n even numbers.

```python
class EvenNumber:
    def __init__(self,n):
        self.n=n
        self.current=0
        self.count=0

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current <self.n:
            result=self.current
            self.current +=2
            self.count +=1
            return result
        else:
            raise StopIteration
        
even_iter=EvenNumber(15)
for num in even_iter:
    print(num)
```

#2. Write a generator that yields the Fibonacci sequence up to n terms.

```python
def fibo(n):
    a=0
    b=1
    for _ in range(n):
        yield a
        a,b=b,a+b

for num in fibo(8):
    print(num)
```

#3. Implement a decorator that prints “Start” before a function and “End” after it.

```python

def timedec(func):
    def wrapper(*args,**kwargs):
        print('start')
        func()
        print('end')
    return wrapper

@timedec
def hello():
    print("hello")
hello()
```

#4. Write a generator to read a CSV file line by line.

```python
def read(file_path):
    with open(file_path,'r') as f:
        for line in f:
            yield line.strip()

for row in read('file.csv'):
    print(row)
```

#5. Implement an iterator class that iterates over a range in reverse order.

```python
class reverseorder:
    def __init__(self,start,end):
        self.current=end-1
        self.start=start

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current >=self.start:
            value=self.current
            self.current -=1
            return value
        else:
            raise StopIteration
        
for num in reverseorder(1,6):
    print(num)
        
        
```
