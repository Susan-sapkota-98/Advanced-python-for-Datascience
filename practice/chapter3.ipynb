#1. Write an iterator that returns the first n even numbers.

```python
class EvenNumber:
    def __init__(self,n):
        self.n=n
        self.current=0
        self.count=0

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current <self.n:
            result=self.current
            self.current +=2
            self.count +=1
            return result
        else:
            raise StopIteration
        
even_iter=EvenNumber(15)
for num in even_iter:
    print(num)
```

#2. Write a generator that yields the Fibonacci sequence up to n terms.

```python
def fibo(n):
    a=0
    b=1
    for _ in range(n):
        yield a
        a,b=b,a+b

for num in fibo(8):
    print(num)
```

#3. Implement a decorator that prints “Start” before a function and “End” after it.

```python

def timedec(func):
    def wrapper(*args,**kwargs):
        print('start')
        func(*args,**kwargs)
        print('end')
    return wrapper

@timedec
def hello():
    print("hello")
hello()
```

#4. Write a generator to read a CSV file line by line.

```python
def read(file_path):
    with open(file_path,'r') as f:
        for line in f:
            yield line.strip()

for row in read('file.csv'):
    print(row)
```

#5. Implement an iterator class that iterates over a range in reverse order.

```python
class reverseorder:
    def __init__(self,start,end):
        self.current=end-1
        self.start=start

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current >=self.start:
            value=self.current
            self.current -=1
            return value
        else:
            raise StopIteration
        
for num in reverseorder(1,6):
    print(num)
        
        
```

#7. Create a decorator to measure memory usage of a function.

```python
import tracemalloc

def memory(func):
    def wrapper(*args, **kwargs):
        tracemalloc.start()  
        
        result = func(*args, **kwargs)
        
        current, peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()    
        
        print(f"{func.__name__} used {peak / 10**6:.6f} MB (peak)")
        
        return result
    return wrapper

@memory
def computesum(n):
    total=sum(range(n))
    return total
computesum(10000)

```

#8. Write a generator that yields prime numbers indefinitely.

```python
def primenum():
    num=2
    while True:
        is_prime=True

        for i in range(2, int(num**0.5)+1):
            if num %i==0:
                is_prime=False
                break
        if is_prime:
            yield num
        num +=1

gen=primenum()
for _ in range(10):
    print(next(gen))
```

#9. Create a decorator to cache results of a function (memoization).

```python
def memoize(func):
    cache={}
    def wrapper(*args,**kwargs):
        if args in cache:
            print("Returning from cache")
            return cache[args]

        result=func(*args)
        cache[args]=result
        return result
    return wrapper

@memoize
def fibo(n):
    if n<=1:
        return n
    return fibo(n-1)+fibo(n-2)
print(fibo(10))
```

#10. Implement a generator that filters only even numbers from a list.

```python
def filter_even(n):
    for nums in n:
        if nums %2==0:
            yield nums

nums=[1,2,3,4,5,6,7,8]

for i in filter_even(nums):
    print(i)
```

#11. Write a generator to simulate a stream of sensor readings.

```python
import random
import time

def sensor_stream():
    while True:
        reading=random.uniform(10,30)
        yield round(reading,2)
        time.sleep(1)

sensor=sensor_stream()

for _ in range(5):
    print(next(sensor))
```

#12. Implement a decorator to log execution time and input arguments.

```python
import time

def logger(func):
    def wrapper(*args,**kwargs):
        print(f'Function Name:{func.__name__}')
        print(f'Arguments:{args}')
        print(f'keyword Arguments: {kwargs}')

        start=time.time()
        result=func(*args,**kwargs)
        end=time.time()

        print(f'Execution Time: {end-start:.6f}')
        print('-'*40)
        return result
    return wrapper

@logger
def add(a,b):
    time.sleep(1)
    return a+b

@logger
def greet(name):
    time.sleep(0.5)
    print(f'Hello {name}')

add(5,10)
greet('susan')
```

#13. Write an iterator that yields only unique items from a list.

```python
class UniqueIterator:
    def __init__(self,data):
        self.data=data
        self.index=0
        self.seen=set()

    def __iter__(self):
        return self
    
    def __next__(self):
        while self.index < len(self.data):
            item=self.data[self.index]
            self.index +=1

            if item not in self.seen:
                self.seen.add(item)
                return item
        raise StopIteration

nums=[1,2,2,3,4,5,5]
unique_iter=UniqueIterator(nums)
for num in unique_iter:
    print(num)
        
```

#14. Implement a decorator to check if inputs of a function are positive numbers.

```python
def positive(func):
    def wrapper(*args, **kwargs):
        
        for arg in args:
            if isinstance(arg, (int, float)) and arg <= 0:
                raise ValueError("All inputs must be positive numbers")

        for value in kwargs.values():
            if isinstance(value, (int, float)) and value <= 0:
                raise ValueError("All inputs must be positive numbers")

        return func(*args, **kwargs)

    return wrapper


@positive
def multiply(a, b):
    return a * b


print(multiply(5, 10))   
# print(multiply(-5, 10)) 
```

#15. Write a generator to simulate an infinite sequence of random numbers.

```python
import random

def infinite_random():
    while True:
        yield random.randint(1,100)
gen=infinite_random()
for _ in range(10):
    print(next(gen))
```
